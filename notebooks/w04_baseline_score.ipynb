## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this import pandas as pd
from pathlib import Path
import pandas as pd


df = pd.read_csv("/content/content_refresh_anonymized.csv")

score = (
    (df["search_volume"] / df["search_volume"].max()) * 40
    + ((1 - df["ctr"]) * 30)
    + (df["days_since_last_update"] / df["days_since_last_update"].max()) * 20
    + (df["trend_direction"] == "down").astype(int) * 10
)

df["baseline_score"] = score

def reason(row):
    if row["ctr"] < 0.03:
        return "LOW_CTR"
    elif row["trend_direction"] == "down":
        return "DECLINING_TREND"
    elif row["days_since_last_update"] > 365:
        return "STALE_CONTENT"
    else:
        return "HIGH_SEARCH_VOLUME"

df["reason_code"] = df.apply(reason, axis=1)
df["action"] = "Review and Refresh"

queue = df.sort_values("baseline_score", ascending=False)

Path("work/outputs").mkdir(parents=True, exist_ok=True)

queue.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

queue.head(20)
```
