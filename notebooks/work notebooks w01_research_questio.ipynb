## 1. My Lane and Why

### Provisional Lane
Content Refresh / Search Ranking

### Why I Selected This Lane

I selected the Content Refresh / Search Ranking lane because it focuses on identifying webpages that should be prioritized for content updates based on their performance. This is a practical machine learning problem where historical search performance data can support better decision-making than simple manual rules. The goal is to help content teams identify declining pages earlier and allocate their efforts more efficiently.

## 2. Research Question

### Research Question

Can machine learning identify webpages that should be prioritized for content refresh more accurately than a manually designed rule?

### Decision

Which webpages should be reviewed first for content updates?

### Action

Prioritize the highest-ranked webpages for content refresh.

### Cost of a Wrong Recommendation

If a declining webpage is not selected, valuable search traffic may continue to decrease.

If a healthy webpage is selected unnecessarily, time and resources may be wasted.

### Why Machine Learning?

Many webpage characteristics influence performance simultaneously. Machine learning can analyze multiple features together and identify useful patterns that may not be obvious through manually designed rules.

```python
import os
import subprocess

if not os.path.exists("flyrank-ml-internship-starter"):
    subprocess.run([
        "git", "clone",
        "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
    ])

os.chdir("flyrank-ml-internship-starter")

print(os.getcwd())




import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

corr = df["search_volume"].corr(df["impressions_90d"])

print("Correlation between search_volume and impressions_90d:", round(corr,3))

median_wc = df.groupby("trend_direction")["word_count"].median()

print("\nMedian Word Count")
print(median_wc.round(0))
```

## 3. Quick Look at the Data

The starter dataset contains:

- **30,000 rows**
- **44 columns**
- **Correlation between search_volume and impressions_90d = 0.001**

The median word count for declining and growing pages is very similar, suggesting that content length alone is not a strong indicator of whether a page is growing or declining.

These observations suggest that search performance depends on multiple factors, making this a suitable machine learning problem rather than a simple rule-based task.
